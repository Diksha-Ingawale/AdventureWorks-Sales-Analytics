```python
for col in numeric_columns:
    Q1 = sales_eda[col].quantile(0.25)
    Q3 = sales_eda[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = sales_eda[
        (sales_eda[col] < lower) |
        (sales_eda[col] > upper)
    ]

    print(f"\nColumn: {col}")
    print("Number of potential outliers:", len(outliers))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "../data/raw/AdventureWorks Sales.xlsx"

data = pd.read_excel(file_path, sheet_name=None)

print("All worksheets loaded successfully!")
print(data.keys())
```

```python
product = data["Product_data"]
date = data["Date_data"]
territory = data["Sales Territory_data"]

print("Product columns:")
print(product.columns.tolist())

print("\nDate columns:")
print(date.columns.tolist())

print("\nTerritory columns:")
print(territory.columns.tolist())
```

```python
# Merge Sales with Product table
sales_product = sales_eda.merge(
    product,
    on="ProductKey",
    how="left"
)

# Merge with Date table
sales_product_date = sales_product.merge(
    date,
    left_on="OrderDateKey",
    right_on="DateKey",
    how="left"
)

# Merge with Territory table
final_sales = sales_product_date.merge(
    territory,
    on="SalesTerritoryKey",
    how="left"
)

print("Combined dataset shape:", final_sales.shape)

display(final_sales.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
top_products = (
    final_sales.groupby("Product")["Sales Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

top_products.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.savefig("../visualizations/chart1_top_products.png", dpi=300)
plt.show()
```

```python
yearly_sales = (
    final_sales.groupby("Fiscal Year")["Sales Amount"]
    .sum()
    .sort_index()
)

plt.figure(figsize=(10, 5))

plt.plot(
    yearly_sales.index.astype(str),
    yearly_sales.values,
    marker="o"
)

plt.title("Yearly Sales Trend")
plt.xlabel("Fiscal Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../visualizations/chart2_yearly_sales.png", dpi=300)
plt.show()
```

```python
territory_sales = (
    final_sales.groupby("Region")["Sales Amount"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

territory_sales.plot(kind="bar")

plt.title("Sales Performance by Territory")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../visualizations/chart3_territory_sales.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=final_sales,
    x="Sales Amount",
    bins=30,
    kde=True
)

plt.title("Distribution of Sales Amount")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("../visualizations/chart4_sales_distribution.png", dpi=300)
plt.show()
```

```python
sample_sales = final_sales.sample(
    n=min(3000, len(final_sales)),
    random_state=42
)

plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=sample_sales,
    x="Order Quantity",
    y="Sales Amount"
)

plt.title("Order Quantity vs Sales Amount")
plt.xlabel("Order Quantity")
plt.ylabel("Sales Amount")
plt.tight_layout()
plt.savefig("../visualizations/chart5_quantity_vs_sales.png", dpi=300)
plt.show()
```
